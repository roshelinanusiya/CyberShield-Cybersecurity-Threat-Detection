```python
from pyspark.ml.feature import StringIndexer, OneHotEncoder, VectorAssembler

# Categorical columns
categorical_cols = ["proto", "service", "state"]

# Numerical columns
exclude_cols = ["id", "attack_cat", "label"] + categorical_cols

numerical_cols = [
    c for c in silver_df.columns
    if c not in exclude_cols
]

print("Numerical features:", len(numerical_cols))
print("Categorical features:", categorical_cols)
```

```python
# Convert categorical values into numerical indices

indexers = [
    StringIndexer(
        inputCol=c,
        outputCol=c + "_index",
        handleInvalid="keep"
    )
    for c in categorical_cols
]

for indexer in indexers:
    silver_df = indexer.fit(silver_df).transform(silver_df)

print("Categorical encoding completed.")
```

```python
from pyspark.ml.feature import StringIndexer, VectorAssembler

# Reload Silver table
silver_df = spark.table(
    "workspace.default.cybersecurity_training_silver"
)

# Categorical columns
categorical_cols = ["proto", "service", "state"]

# Exclude non-feature columns
exclude_cols = ["id", "attack_cat", "label"] + categorical_cols

# Numerical columns
numerical_cols = [
    c for c in silver_df.columns
    if c not in exclude_cols
]

# Encode categorical columns
encoded_df = silver_df

for c in categorical_cols:
    indexer = StringIndexer(
        inputCol=c,
        outputCol=c + "_index",
        handleInvalid="keep"
    )
    encoded_df = indexer.fit(encoded_df).transform(encoded_df)

# Combine numerical + encoded categorical features
feature_cols = numerical_cols + [
    c + "_index" for c in categorical_cols
]

assembler = VectorAssembler(
    inputCols=feature_cols,
    outputCol="features",
    handleInvalid="keep"
)

gold_df = assembler.transform(encoded_df)

# Keep only ML-ready columns
gold_df = gold_df.select("features", "label")

print("Gold dataset created successfully!")
print("Number of rows:", gold_df.count())

display(gold_df.limit(10))
```

```python
gold_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable(
        "workspace.default.cybersecurity_training_gold"
    )

print("Gold table saved successfully!")
```

```python
# Load Gold dataset

gold_df = spark.table(
    "workspace.default.cybersecurity_training_gold"
)

# Split into training and validation data
train_df, validation_df = gold_df.randomSplit(
    [0.8, 0.2],
    seed=42
)

print("Training records:", train_df.count())
print("Validation records:", validation_df.count())
```

```python
from pyspark.ml.classification import RandomForestClassifier

rf = RandomForestClassifier(
    featuresCol="features",
    labelCol="label",
    numTrees=100,
    maxDepth=10,
    maxBins=200,
    seed=42
)

rf_model = rf.fit(train_df)

print("Random Forest training completed!")
```

```python
predictions = rf_model.transform(validation_df)

display(
    predictions.select(
        "label",
        "prediction",
        "probability"
    ).limit(20)
)
```

```python
from pyspark.ml.evaluation import MulticlassClassificationEvaluator

evaluator = MulticlassClassificationEvaluator(
    labelCol="label",
    predictionCol="prediction",
    metricName="accuracy"
)

accuracy = evaluator.evaluate(predictions)

print(f"Model Accuracy: {accuracy * 100:.2f}%")
```

```python
predictions = rf_model.transform(validation_df)

display(
    predictions.select(
        "label",
        "prediction",
        "probability"
    ).limit(20)
)
```

```python
from pyspark.ml.evaluation import MulticlassClassificationEvaluator

metrics = {}

for metric in ["accuracy", "weightedPrecision", "weightedRecall", "f1"]:
    evaluator = MulticlassClassificationEvaluator(
        labelCol="label",
        predictionCol="prediction",
        metricName=metric
    )
    
    metrics[metric] = evaluator.evaluate(predictions)

print(f"Accuracy  : {metrics['accuracy']:.4f}")
print(f"Precision : {metrics['weightedPrecision']:.4f}")
print(f"Recall    : {metrics['weightedRecall']:.4f}")
print(f"F1 Score  : {metrics['f1']:.4f}")
```

```python
confusion_matrix = (
    predictions
    .groupBy("label", "prediction")
    .count()
    .orderBy("label", "prediction")
)

display(confusion_matrix)
```
